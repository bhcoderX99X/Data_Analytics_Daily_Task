```python
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
x=[2,3,1,4,2]
y=[100,200,100,150,200]

sns.lineplot(x)
plt.title("Sales")
plt.xlabel("days")
plt.ylabel("no. of occupencies")
plt.show()
```

```python
sns.barplot(x)
plt.show()
```

```python
sns.kdeplot(x)     #kde stands for kernel density estimation
plt.show()
```

```python
# kde + boxplot = violinplot -> dono ek saath dekhni ho tab violinplot ka use karte hai

sns.violinplot(x)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_excel("IIOT_Dataset.xlsx")
```

```python
sns.histplot(data=df, x="Revenue", bins=20)  
plt.show()

```

```python
Q1=df['Unit_Price'].quantile(0.25)
Q3=df['Unit_Price'].quantile(0.75)
IQR=Q3-Q1

lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
outliers=df[(df['Unit_Price']< lower_bound) | (df['Unit_Price'] > upper_bound)]

outliers
```
