```python
# ARRAY PROPERTIES

arr = np.array([[10,20,30], [40,50,60]])

print("Array Dimension :",arr.ndim)
print("Row and Column :", arr.shape)
print("Total number of element :", arr.size)
print("Data Type of element :", arr.dtype)
```

```python
# INDEXING AND SLICING

a = np.array([10, 20, 30, 40, 50, 60, 70])

print(a[2:6])
```

```python
a = np.array([[10, 20, 30], 
              [50, 60, 70], 
              [40, 80, 90]])
print(a)

print(a[0,1])
print(a[:, 1])                # All Rows Column 1
print(a[0:2, 1:3])            # Top right 2x2 sub-matrix
```

```python
x = np.array([1,2,3])
y = np.array([10,20,30])

print(x+y)
print(x*2)
print(x**2)

print(np.sin(x))
print(np.exp)
```

```python
# Reshaping 1D (6 elements) into 2D (2x3)

flat = np.array([1, 2, 3, 4, 5, 6])
grid = flat.reshape(2, 3)

data = np.array([[1, 2], [3, 4]])

print(np.sum(data))                  # Total sum
print(np.sum(data, axis=0))          # Sum down columns
print(np.sum(data, axis=1))          # Sum across rows
print(np.mean(data))      
print(np.max(data)) 
```

```python
matrix = np.array([[1,1,1],         
                   [2,2,2]])            # Shape = (2,3)
vector = np.array([10,20,30])           # Shape = (3,)
                                        # 
result = matrix + vector

print(result)
```

```python
# Create a 2D NumPy array with 3 rows and 4 columns containing numbers from 10 to 21.
# Print its shape, number of dimensions, and total size.

import numpy as np

# Create the array using arange and reshape
arr = np.arange(10, 22).reshape(3, 4)
print("Array:\n", arr)

print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Total elements:", arr.size)
```
