```python
# Revenue Statistics

print("Total Revenue :", df['Revenue'].sum())
print("Average Revenue :", df['Revenue'].mean())
print("Maximum Revenue :", df['Revenue'].max())
print("Minimum Revenue :", df['Revenue'].min())
```

## Sales Analytics

```python
# Total Sales KPI

total_revenue = df['Revenue'].sum()
total_profit = df['Profit'].sum()
total_units = df['Quantity_Sold'].sum()

print(total_revenue)
print(total_profit)
print(total_units)
```

```python
# Monthly Revenue Trend

monthly_sales = (
    df.groupby(df['Date'].dt.month)
      ['Revenue']
      .sum()
      .reset_index()
)

monthly_sales.columns = ['Month','Revenue']

print(monthly_sales)
```

```python
# Monthly Profit Trend

monthly_profit = (
    df.groupby(df['Date'].dt.month)
      ['Profit']
      .sum()
      .reset_index()
)

print(monthly_profit)
```

```python
# Region Wise Revenue

region_sales = (
    df.groupby('Region')
      ['Revenue']
      .sum()
      .sort_values(ascending=False)
)

print(region_sales)
```

```python
# Store Wise Revenue

store_sales = (
    df.groupby('Store_Location')
      ['Revenue']
      .sum()
      .sort_values(ascending=False)
)

print(store_sales)
```

```python
# Customer Segmentation Analytics Customers by Gender

gender_count = df['Gender'].value_counts()
print(gender_count)
```

```python
# Revenue by Gender

gender_revenue = (
    df.groupby('Gender')
      ['Revenue']
      .sum()
)
print(gender_revenue)
```

```python
#Revenue by Customer Segment

segment_revenue = (
    df.groupby('Segment')
      ['Revenue']
      .sum()
      .sort_values(ascending=False)
)

print(segment_revenue)
```

```python
# Profit by Customer Segment

segment_profit = (
    df.groupby('Segment')
      ['Profit']
      .sum()
      .sort_values(ascending=False)
)

print(segment_profit)
```

```python
# Top 10 Customers

top_customers = (
    df.groupby('Customer_ID')
      ['Revenue']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_customers)
```

```python
# Product Performance Analytics Revenue by Category

category_sales = (
    df.groupby('Product_Category')
      ['Revenue']
      .sum()
      .sort_values(ascending=False)
)

print(category_sales)
```

```python
# Profit by Category

category_profit = (
    df.groupby('Product_Category')
      ['Profit']
      .sum()
      .sort_values(ascending=False)
)

print(category_profit)
```

```python
# Top Selling Products

top_products = (
    df.groupby('Product_Name')
      ['Quantity_Sold']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_products)
```

```python
# Most Profitable Products

top_profit_products = (
    df.groupby('Product_Name')
      ['Profit']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_profit_products)
```

```python
# Store Performance Analytics - Revenue by Store

store_performance = (
    df.groupby('Store_Location')
      .agg(
            Total_Revenue=('Revenue','sum'),
            Total_Profit=('Profit','sum'),
            Quantity_Sold=('Quantity_Sold','sum')
      )
      .sort_values('Total_Revenue', ascending=False)
)

print(store_performance)
```

```python
# Average Revenue Per Store

avg_store_revenue = (
    df.groupby('Store_Location')
      ['Revenue']
      .mean()
)

print(avg_store_revenue)
```

```python
# Inventory Analytics Total Quantity Sold By Product

inventory = (
    df.groupby('Product_Name')
      ['Quantity_Sold']
      .sum()
      .sort_values(ascending=False)
)

print(inventory)
```

```python
# Fast Moving Products
fast_moving = inventory.head(10)

print(fast_moving)
```

```python
# Slow Moving Products

slow_moving = inventory.tail(10)
print(slow_moving)
```

```python
# Profitability Analysis
# Profit Margin %

df['Profit_Margin_%'] = (
    df['Profit'] /
    df['Revenue']
) * 100

print(df[['Revenue','Profit','Profit_Margin_%']].head())
```

```python
# Average Profit Margin

print(df['Profit_Margin_%'].mean())
```

```python
# Category Profit Margin

category_margin = (
    df.groupby('Product_Category')
      ['Profit_Margin_%']
      .mean()
      .sort_values(ascending=False)
)

print(category_margin)
```

```python
#Time Series Analysis
# Sales by Month

df['Month'] = df['Date'].dt.month
sales_trend = (
    df.groupby('Month')
      ['Revenue']
      .sum()
      .reset_index()
)
print(sales_trend)
```

```python
# Sales by Quarter

df['Quarter'] = df['Date'].dt.quarter

quarter_sales = (
    df.groupby('Quarter')
      ['Revenue']
      .sum()
)

print(quarter_sales)
```

```python
# Executive Dashboard KPIs

dashboard = {
    "Total Revenue": df['Revenue'].sum(),
    "Total Profit": df['Profit'].sum(),
    "Total Units Sold": df['Quantity_Sold'].sum(),
    "Average Revenue": round(df['Revenue'].mean(),2),
    "Average Profit": round(df['Profit'].mean(),2),
    "Unique Customers": df['Customer_ID'].nunique(),
    "Unique Products": df['Product_ID'].nunique(),
    "Stores": df['Store_Location'].nunique()
}

for k,v in dashboard.items():
    print(f"{k}: {v}")
```

```python
# Final Summary DataFrame

summary = pd.DataFrame({
    "Metric":[
        "Total Revenue",
        "Total Profit",
        "Total Quantity Sold",
        "Average Revenue",
        "Average Profit"
    ],
    "Value":[
        df['Revenue'].sum(),
        df['Profit'].sum(),
        df['Quantity_Sold'].sum(),
        df['Revenue'].mean(),
        df['Profit'].mean()
    ]
})

print(summary)
```

```python
# Monthly Sales report

monthly_sales.loc[
    monthly_sales['Revenue'].idxmax()
]
```

```python
weekday_sales = (
    df.groupby('Day_Name', as_index=False)
      .agg(
          Revenue=('Revenue','sum'),
          Profit=('Profit','sum')
      )
)

order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

weekday_sales['Day_Name'] = pd.Categorical(
    weekday_sales['Day_Name'],
    categories=order,
    ordered=True
)

weekday_sales = weekday_sales.sort_values('Day_Name')

print(weekday_sales)
```

```python
#Month-over-Month Growth (MoM)

mom_sales = (
    df.groupby(['Year','Month'])
      ['Revenue']
      .sum()
      .reset_index()
)
mom_sales['MoM_Growth_%'] = (
    mom_sales['Revenue']
    .pct_change()
    * 100
)
print(mom_sales)
```

```python
# Revenue Heatmap Table

sales_heatmap = pd.pivot_table(
    df,
    values='Revenue',
    index='Month_Name',
    columns='Region',
    aggfunc='sum',
    fill_value=0
)

print(sales_heatmap)
```

```python
# Seasonal Product Performance

season_product = (
    df.groupby(
        ['Quarter','Product_Category']
    )['Revenue']
    .sum()
    .reset_index()
)

print(season_product)
```

```python
# Store Performance Over Time

store_monthly = (
    df.groupby(
        ['Store_Location',
         'Month_Name']
    )['Revenue']
    .sum()
    .reset_index()
)

print(store_monthly)
```
